## The Phase 1 gate (Friday demo with mentor)

Your mentor will ask you to:

1. **Fresh install demo:** clone english-helper repo on the mentor's machine (or in a fresh dir). `uv tool install --from . english-helper`. Run `english-helper add ubiquitous`. Show the result. Then same for pomo.

2. **Walk through the architecture:** explain how english-helper's 5 modules relate. Why each is separate. What a new feature (e.g., `audio-play` command) would look like — which files would change.

3. **Random code walkthrough:** the mentor picks a file at random (e.g., `scheduler.py`). You explain what it does, why it's structured this way, and what would break if you deleted the third method.

4. **Live coding (45 min):** add a small feature to english-helper. Mentor picks the feature. Examples:
   - `pomo today` — shortcut for `list --since today`
   - `english-helper top` — show the top 5 most-quizzed words by review count
   - `english-helper export FILE` — dump vocab as CSV
   
   Acceptance: works end-to-end, has tests, passes mypy + ruff, you ship a PR.

5. **The retrospective conversation:** mentor reads your retro doc. Asks one or two follow-ups.
