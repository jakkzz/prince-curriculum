## Section B — Predict the output

**B1.**

```python
from dataclasses import dataclass, field
@dataclass
class Store:
    items: dict = field(default_factory=dict)
    def add(self, k, v): self.items[k] = v
s = Store(); s.add('a', 1); print(s.items)
```

**B2.**

```python
@dataclass
class S:
    items: dict = field(default_factory=dict)
s1 = S(); s2 = S(); s1.items['x'] = 1; print(s2.items)
```

**B3.**

```python
@dataclass
class Q:
    rounds: int = 5
print(Q().rounds)
```

**B4.**

```python
@dataclass(frozen=True)
class W:
    word: str
w = W('x')
w.word = 'y'
```

**B5.**

```python
from abc import ABC, abstractmethod
class M(ABC):
    @abstractmethod
    def f(self): ...
class R(M):
    def f(self): return 1
print(R().f())
```

**B6.**

```python
class C:
    def __init__(self): self.x = 1
    @property
    def doubled(self): return self.x * 2
c = C(); c.x = 5; print(c.doubled)
```

**B7.**

```python
from dataclasses import dataclass, asdict
@dataclass
class P:
    x: int; y: int
print(asdict(P(1, 2)))
```

**B8.**

```python
from typing import Protocol
class HasSpeak(Protocol):
    def speak(self) -> str: ...
def shout(x: HasSpeak): print(x.speak())
class Dog:
    def speak(self): return 'woof'
shout(Dog())
```
