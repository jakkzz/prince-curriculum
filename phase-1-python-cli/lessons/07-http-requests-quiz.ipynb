## Section B — Predict the output

**B1.**

```python
import requests
r = requests.get('https://httpbin.org/status/200', timeout=5)
print(r.status_code)
print(r.ok)
```

**B2.**

```python
r = requests.get('https://httpbin.org/status/404', timeout=5)
print(r.status_code)
print(r.ok)
```

**B3.**

```python
r = requests.get('https://httpbin.org/json', timeout=5)
data = r.json()
print(type(data))
```

**B4.**

```python
url = 'https://api.dictionaryapi.dev/api/v2/entries/en/cat'
r = requests.get(url, timeout=5)
data = r.json()
print(type(data))
```

**B5.**

```python
r = requests.get('https://httpbin.org/get?word=thorough', timeout=5)
print(r.json()['args'])
```

**B6.**

```python
r = requests.get('https://httpbin.org/headers', timeout=5)
print(r.json()['headers']['Host'])
```

**B7.**

```python
headers = {'X-Custom': 'hello'}
r = requests.get('https://httpbin.org/headers', headers=headers, timeout=5)
print(r.json()['headers']['X-Custom'])
```

**B8.**

```python
r = requests.get('https://httpbin.org/status/500', timeout=5)
try:
    r.raise_for_status()
    print('ok')
except requests.HTTPError as e:
    print('error', r.status_code)
```
