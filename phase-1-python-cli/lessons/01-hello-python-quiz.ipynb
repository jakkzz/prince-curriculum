**A1.** What does the `=` operator do in Python?

A) Compares two values for equality
B) Assigns a value to a name
C) Defines a function
D) Calls a method

<details>
<summary>Answer</summary>

**B**

Assignment. `age = 16` means "the name `age` now refers to the value 16." For equality comparison, use `==`.
</details>

**A2.** Which of these is NOT a valid Python variable name?

A) `my_age`
B) `_count`
C) `2nd_place`
D) `firstName`

<details>
<summary>Answer</summary>

**C** — `2nd_place`

Python variable names must start with a letter or underscore, never a digit. `my_age` and `_count` are fine. `firstName` is valid but unconventional in Python (prefer `first_name`).
</details>

**A3.** What's the type of the value `3.0`?

A) `int`
B) `float`
C) `str`
D) `decimal`

<details>
<summary>Answer</summary>

**B** — `float`

The presence of the decimal point makes it a float, even though the value happens to be a whole number. `3` is an int; `3.0` is a float.
</details>

**A4.** What does this expression evaluate to? `7 % 2`

A) 3.5
B) 3
C) 1
D) 0

<details>
<summary>Answer</summary>

**C** — `1`

`%` is the modulo operator: "the remainder of dividing left by right." 7 divided by 2 is 3, remainder 1.
</details>

**A5.** Which f-string is valid?

A) `f"My name is {name}"`
B) `"My name is {name}".f()`
C) `"f My name is {name}"`
D) `"My name is $name"`

<details>
<summary>Answer</summary>

**A**

The `f` goes immediately before the opening quote. Inside the string, `{...}` lets you embed any Python expression. Other languages use different syntax (Bash uses `$name`, JavaScript uses `${name}`), but in Python it's `f"...{expr}..."`.
</details>

**A6.** Which is an expression?

A) `x = 5`
B) `print("hi")`
C) `2 + 3`
D) `if x: pass`

<details>
<summary>Answer</summary>

**C** — `2 + 3`

It evaluates to a value (5). Statements (`x = 5`, `print(...)`, `if`) do something but don't produce a value usable on the right side of `=`. Tricky case: `print("hi")` is a function call, which IS technically an expression (returns `None`), but in the context of "expression vs statement," the canonical answer here is `2 + 3` — purely math, clearly a value.
</details>

**A7.** What does `2 ** 10` evaluate to?

A) 20
B) 12
C) 1024
D) 100

<details>
<summary>Answer</summary>

**C** — `1024`

`**` is the exponent operator. `2 ** 10` = 2 to the 10th power = 1024. Memorize this one — it shows up everywhere in computing.
</details>

**A8.** What does `int("42")` evaluate to?

A) `42` (an int)
B) `"42"` (a string)
C) `42.0` (a float)
D) Error — can't convert

<details>
<summary>Answer</summary>

**A** — `42` as an int

`int(...)` converts a string to an int (if it represents a valid whole number). This is what you use after `input()`, which always returns a string.
</details>

**A9.** What's the difference between `/` and `//` in Python?

A) They do the same thing
B) `/` always returns a float; `//` returns an int (floor division)
C) `/` is integer division; `//` is float division
D) `//` is a comment

<details>
<summary>Answer</summary>

**B**

`/` is true division — always returns a float (`7/2` = `3.5`, `6/2` = `3.0`). `//` is floor division — drops the fractional part (`7//2` = `3`). `//` is NOT a comment in Python (that's `#`).
</details>

**A10.** A Jupyter notebook cell ends with this line: `result`. What happens?

A) Nothing — only `print()` causes output
B) The notebook displays the value of `result`
C) Error — bare names aren't allowed
D) It saves `result` to disk

<details>
<summary>Answer</summary>

**B**

Notebooks automatically display the value of the **last expression** in a cell. No `print()` needed — just put the expression on the last line. This is unique to notebooks; in a `.py` file the bare expression would do nothing.
</details>

````python
n = 0.1 + 0.2
print(n == 0.3)
````

<details>
