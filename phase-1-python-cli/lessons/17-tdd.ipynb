## The TDD cycle

```
1. RED:     Write a failing test for behavior that doesn't exist
2. GREEN:   Write the smallest code to make it pass
3. REFACTOR: Clean up. Tests still green.
→ go to 1
```

**Critical:** see the test fail first. If a test passes without your implementation, the test is buggy.

## Walkthrough — adding `WordStore.search(pattern)`

Goal: a method that returns words containing a substring. Doesn't exist yet.

### Red

````python
# tests/test_storage.py
def test_search_finds_substring():
    store = WordStore()
    store.add(Word('thorough'))
    store.add(Word('nuance'))
    store.add(Word('cat'))
    results = store.search('th')
    assert [w.word for w in results] == ['thorough']
````

Run: `pytest -v tests/test_storage.py::test_search_finds_substring` — fails (AttributeError: no method `search`).

### Green — minimum impl

````python
# storage.py
def search(self, pattern: str) -> list[Word]:
    return [w for w in self.words.values() if pattern in w.word]
````

Run: passes. ✅

### Red — next test

````python
def test_search_case_insensitive():
    store = WordStore()
    store.add(Word('Thorough'))
    results = store.search('THOROUGH')
    assert len(results) == 1
````

Fails: substring match is case-sensitive. Update implementation.

````python
def search(self, pattern: str) -> list[Word]:
    p = pattern.lower()
    return [w for w in self.words.values() if p in w.word.lower()]
````

Both tests pass. ✅

### Refactor (optional)

Code is clean. Move on.

**Total time:** maybe 10 minutes for a real feature, with full test coverage. **That's the magic.**

## Why TDD works

1. **Forces you to design the API.** Before writing implementation, you decided what `search()` should look like to callers.
2. **Catches over-engineering.** You only write code that's needed to make a test pass.
3. **Confident refactors.** You can rip and rewrite knowing tests prove behavior unchanged.
4. **Honest documentation.** Tests describe what the code does — and they can't lie (or they fail).

> 💡 **In the wild:** Most teams don't TDD everything (slow learning curve, sometimes overhead-heavy for trivial functions). But TDD is the gold standard for non-trivial features. Try it on every Phase 1 Week 4 task.
