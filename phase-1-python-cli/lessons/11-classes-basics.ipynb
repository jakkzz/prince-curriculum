## The mental model

A class is a blueprint. An instance is one specific thing built from it.

```
Word                      ← class (blueprint)
├─ word: str
├─ ipa: str
└─ quiz_me() method

Word('thorough', '/x/')   ← instance #1
Word('nuance', '/y/')     ← instance #2
```

Each instance has its OWN data. Methods are shared (defined once).

## 1. Define and instantiate

````python
class Word:
    def __init__(self, word: str, ipa: str):
        self.word = word
        self.ipa = ipa

w = Word('thorough', '/ˈθʌrə/')
f"{w.word} {w.ipa}"
````

**Anatomy:**

- `class Word:` — declares a new type
- `def __init__(self, ...):` — the constructor. Called on `Word(...)`.
- `self` — the instance being constructed. Just a parameter; could be `me` or `this` (don't).
- `self.word = word` — stores `word` as an attribute of this instance.
- `Word('thorough', '/x/')` — creates an instance, returns it.

## 2. Methods — functions inside a class

````python
class Word:
    def __init__(self, word: str, ipa: str):
        self.word = word
        self.ipa = ipa

    def quiz_me(self, guess: str) -> bool:
        return guess.strip() == self.ipa

    def shout(self) -> str:
        return self.word.upper() + '!'

w = Word('thorough', '/ˈθʌrə/')
print(w.quiz_me('/ˈθʌrə/'))   # True
print(w.shout())              # THOROUGH!
````

When you call `w.quiz_me('/x/')`, Python translates to `Word.quiz_me(w, '/x/')`. The `self` parameter receives the instance. **Always include `self` as the first parameter of any method.**

## 3. The dunder methods you'll actually use

````python
class Word:
    def __init__(self, word: str, ipa: str):
        self.word = word
        self.ipa = ipa

    def __repr__(self) -> str:
        return f'Word({self.word!r}, {self.ipa!r})'

    def __eq__(self, other: object) -> bool:
        if not isinstance(other, Word):
            return NotImplemented
        return self.word == other.word and self.ipa == other.ipa

    def __hash__(self) -> int:
        return hash((self.word, self.ipa))

w1 = Word('thorough', '/ˈθʌrə/')
w2 = Word('thorough', '/ˈθʌrə/')
print(w1)            # uses __repr__
print(w1 == w2)      # True (custom __eq__)
print({w1, w2})      # set of 1 — __hash__ makes them equal
````

Methods starting with `__` and ending with `__` are **dunder** (double-underscore). They hook into Python's syntax:

- `__init__` — constructor
- `__repr__` — `repr(obj)` and the REPL display
- `__str__` — `str(obj)` and `print(obj)`; falls back to `__repr__` if missing
- `__eq__` — `obj == other`; without this, default is `is` (identity)
- `__hash__` — required for set/dict keys; pair with `__eq__`
- `__len__` — `len(obj)`
- `__iter__` — `for x in obj:`
