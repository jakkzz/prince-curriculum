## Section A — Multiple choice

1 point each.

**A1.** Which of these is mutable?

A) `tuple`
B) `frozenset`
C) `list`
D) `str`

<details>
<summary>Answer</summary>

**C** — `list`

Lists, dicts, sets are mutable. Tuples, frozensets, strings, ints, floats are immutable. "Mutable" means "you can change its contents in place" — `xs.append(99)` works on a list but not a tuple.
</details>

**A2.** What's the type of `True + 1`?

A) `bool`
B) `int`
C) `str`
D) Error

<details>
<summary>Answer</summary>

**B** — `int`

In Python, `bool` is a subclass of `int`. `True` behaves like `1`, `False` like `0`. So `True + 1 == 2` and `type(True + 1)` is `int`. This is technically clever, practically rarely used.
</details>

**A3.** Which type would you choose for storing a person's age?

A) `int`
B) `float`
C) `str`
D) `bool`

<details>
<summary>Answer</summary>

**A** — `int`

Age is a whole number. Floats introduce imprecision; strings can't be compared numerically. `int` is the right choice.
</details>

**A4.** Which is the modern, recommended type-hint syntax for "a string or None"?

A) `Optional[str]`
B) `str | None`
C) `str?`
D) `Maybe[str]`

<details>
<summary>Answer</summary>

**B** — `str | None`

Both A and B work (B requires Python 3.10+). `str | None` is the modern Python style — shorter and matches the type-theoretic notation. `Optional[str]` is older and requires `from typing import Optional`.
</details>

**A5.** Which lookup is fastest for large collections?

A) `x in [1, 2, 3, ..., 1_000_000]` (list)
B) `x in (1, 2, 3, ..., 1_000_000)` (tuple)
C) `x in {1, 2, 3, ..., 1_000_000}` (set)
D) All the same

<details>
<summary>Answer</summary>

**C** — set

Set/dict lookups are O(1) — instant regardless of size, because they use a hash table. List/tuple lookups are O(n) — they scan from the start. For "is this value in my collection?" with many values, **always use a set or dict.**
</details>

**A6.** What does `len({"a", "a", "b", "c", "b"})` evaluate to?

A) 5
B) 3
C) 2
D) Error

<details>
<summary>Answer</summary>

**B** — 3

Sets drop duplicates automatically. `{"a", "a", "b", "c", "b"}` becomes `{"a", "b", "c"}` — 3 unique items. This is a great trick for deduplication: `list(set(my_list))`.
</details>

**A7.** Which of these is the **falsy** value group?

A) `0`, `""`, `[]`, `None`, `False`
B) `0`, `" "`, `[0]`, `None`, `False`
C) `1`, `"hello"`, `[1]`, `None`, `False`
D) `0.0`, `"0"`, `[]`, `None`, `True`

<details>
<summary>Answer</summary>

**A**

Falsy values: `False`, `None`, `0`, `0.0`, `""`, `[]`, `()`, `{}`, `set()`. Notice `"0"` is NOT falsy — it's a non-empty string. `[0]` is NOT falsy — it's a non-empty list (containing a falsy value, but the list itself is truthy).
</details>

**A8.** Why use `is None` instead of `== None`?

A) `==` is slower
B) `is` checks object identity; safer for `None` (a singleton)
C) `is None` is a different language feature
D) `== None` doesn't exist

<details>
<summary>Answer</summary>

**B**

`is` checks "same object in memory." `==` calls `.__eq__()`, which a malicious or buggy class could override. `None` is a singleton (only one ever exists), so `is None` is the idiomatic, unambiguous, slightly faster check.
</details>

**A9.** What's the type of `(1,)` vs `(1)`?

A) Both are tuples
B) Both are ints
C) `(1,)` is a tuple; `(1)` is an int (parens are just grouping)
D) `(1)` is a tuple; `(1,)` is an int

<details>
<summary>Answer</summary>

**C**

This is a classic Python gotcha. A 1-element tuple needs the trailing comma: `(1,)`. `(1)` is just `1` with redundant parens. `()` is the empty tuple. For 2+ elements, the comma is implicit: `(1, 2)`.
</details>

**A10.** Which is the right type hint for "a dictionary mapping strings to lists of integers"?

A) `dict<str, list<int>>`
B) `Dict[String, List[Integer]]`
C) `dict[str, list[int]]`
D) `{str: [int]}`

<details>
<summary>Answer</summary>

**C** — `dict[str, list[int]]`

Python generic types use `[]` not `<>`. Use lowercase built-in names (modern, since Python 3.9): `dict`, `list`, `set`, `tuple`. The older `Dict[..., ...]` from `typing` still works but isn't needed in new code.
</details>

## Section B — Predict the output

1 point each. Predict before running.

**B1.**

````python
x = [1, 2, 3]
y = x
y.append(4)
print(x)
````

<details>
<summary>Expected output</summary>

`[1, 2, 3, 4]`

**Aliasing!** Lists are mutable. `y = x` makes `y` point to the **same list**. Modifying through one name affects the other. To copy, use `y = x.copy()` or `y = list(x)`.
</details>

**B2.**

````python
a = (1, 2, 3)
b = a
b = (4, 5, 6)
print(a)
````

<details>
<summary>Expected output</summary>

`(1, 2, 3)`

Different from B1 because tuples are **immutable**. `b = (4, 5, 6)` doesn't modify the original tuple — it rebinds `b` to a new tuple. `a` still points to the original.
</details>

**B3.**

````python
words = {"apple", "banana", "apple", "cherry"}
print(len(words))
````

<details>
<summary>Expected output</summary>

`3`

Sets drop duplicates. `"apple"` only counts once.
</details>

**B4.**

````python
user = {"name": "Prince", "age": 16}
print(user.get("email", "no email"))
````

<details>
<summary>Expected output</summary>

`no email`

`dict.get(key, default)` returns the default if the key isn't there. Safer than `user["email"]` which raises `KeyError`.
</details>

**B5.**

````python
x: int = 5
y: int = "hello"
print(type(x), type(y))
````

<details>
<summary>Expected output</summary>

`<class 'int'> <class 'str'>`

**Type hints are not enforced at runtime.** Python lets you assign a string to a variable hinted as `int`. mypy would complain; the interpreter doesn't. The actual type is determined by the assigned value.
</details>

**B6.**

````python
name = None
if name:
    print("have a name")
else:
    print("no name")
````

<details>
<summary>Expected output</summary>

`no name`

`None` is falsy. The `if` branch doesn't execute. Same result for empty string `""`, empty list `[]`, `0`.
</details>

**B7.**

````python
if 0:
    print("zero is truthy")
else:
    print("zero is falsy")
````

<details>
<summary>Expected output</summary>

`zero is falsy`

`0` is falsy. So is `0.0`, `False`, `None`, `""`, `[]`, `{}`, `set()`.
</details>

**B8.**

````python
user_input = "42"
result = int(user_input) * 2
print(result, type(result))
````

<details>
<summary>Expected output</summary>

`84 <class 'int'>`

`input()` returns a string. `int("42")` converts to 42. `42 * 2 = 84`. Result is an int.
</details>

## Section D — Find the bug

1 point each.

**D1.**

````python
user = {"name": "Prince", "age": 16}
print(user.email)
````

<details>
<summary>Diagnosis + fix</summary>

**Error:** `AttributeError: 'dict' object has no attribute 'email'`

**Fix:** Dicts use `["key"]` not `.attribute`. `print(user["email"])` — but that'll raise `KeyError`. Safer: `print(user.get("email", "not set"))`.
</details>

**D2.**

````python
nums = [1, 2, 3]
nums = nums.append(4)
print(nums)
````

<details>
<summary>Diagnosis + fix</summary>

**Output:** `None`

**Bug:** `.append()` modifies the list **in place** and returns `None`. Reassigning erases your list. **Fix:** Just `nums.append(4)` (no `nums = `).
</details>

**D3.**

````python
age: int = 16
age = "sixteen"
print(age + 1)
````

<details>
<summary>Diagnosis + fix</summary>

**Runtime error:** `TypeError: can only concatenate str (not "int") to str`

**Bug:** The type hint said `int`, but at runtime nothing enforces it. The reassignment to `"sixteen"` works. Then `+ 1` fails. **Lesson:** type hints are checked by mypy, not at runtime. Run `uv run mypy file.py` to catch this.
</details>

**D4.**

````python
x = None
if x == None:
    print("x is None")
````

<details>
<summary>Diagnosis + fix</summary>

**Not technically an error** but a style/correctness issue. Linters flag `== None`. Use `if x is None:` — the idiomatic, faster, and unambiguous form.
</details>

**D5.**

````python
items = (1, 2, 3)
items.append(4)
````

<details>
<summary>Diagnosis + fix</summary>

**Error:** `AttributeError: 'tuple' object has no attribute 'append'`

**Fix:** Tuples are immutable. If you need to add items, you wanted a list. Change `()` to `[]`, OR if you really need a tuple: `items = items + (4,)` (creates a new tuple).
</details>

**E6.** Write a typed function `count_long_words(words, min_length=8)` that returns how many words in the list are at least `min_length` letters long.

````python
# Your code
````

<details>
<summary>Solution</summary>

```python
def count_long_words(words: list[dict[str, str]], min_length: int = 8) -> int:
    count = 0
    for entry in words:
        if len(entry["word"]) >= min_length:
            count += 1
    return count
```

**Pythonic one-liner (after Day 5):** `sum(1 for e in words if len(e['word']) >= min_length)`.
</details>

**E7.** Write a typed function `find_word(words, target)` that returns the word entry matching `target`, or `None` if not found. Use `dict[str, str] | None` as the return type.

````python
# Your code
````

<details>
<summary>Solution</summary>

```python
def find_word(words: list[dict[str, str]], target: str) -> dict[str, str] | None:
    for entry in words:
        if entry["word"] == target:
            return entry
    return None
```
