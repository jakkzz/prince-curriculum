## Section B — Predict the output

**B1.**

```python
import argparse
p = argparse.ArgumentParser()
p.add_argument('name')
args = p.parse_args(['Prince'])
print(args.name)
```

**B2.**

```python
import argparse
p = argparse.ArgumentParser()
p.add_argument('--age', type=int, default=0)
args = p.parse_args(['--age', '16'])
print(args.age, type(args.age))
```

**B3.**

```python
import argparse
p = argparse.ArgumentParser()
p.add_argument('-v', '--verbose', action='store_true')
args = p.parse_args([])
print(args.verbose)
args = p.parse_args(['-v'])
print(args.verbose)
```

**B4.**

```python
import argparse
p = argparse.ArgumentParser()
p.add_argument('name', nargs='?', default='world')
print(p.parse_args([]).name)
print(p.parse_args(['Prince']).name)
```

**B5.**

```python
import argparse
p = argparse.ArgumentParser()
p.add_argument('words', nargs='+')
print(p.parse_args(['a', 'b', 'c']).words)
```

**B6.**

```python
import argparse
p = argparse.ArgumentParser()
p.add_argument('--mode', choices=['random', 'sequential'])
p.parse_args(['--mode', 'random'])
p.parse_args(['--mode', 'invalid'])
```
