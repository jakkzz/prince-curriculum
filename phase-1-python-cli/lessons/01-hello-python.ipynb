## 2. Variables

A variable is a name attached to a value. The `=` is not "equals" — it's "this name now refers to this value."

````python
age = 16
age
````

````python
next_year = age + 1
next_year
````

> ⚠️ **One name per concept.** Don't reuse `age` to mean different things in different places. Read your future self's code: `age` should always mean age. If you have a new concept, give it a new name.

## 3. Strings — and the magic of f-strings

Strings are text, wrapped in quotes. Double or single, your choice — be consistent.

````python
name = "Prince"
greeting = "Hello, " + name + "!"
greeting
````

That `+` concatenation works but it's ugly. The modern way is an **f-string** — an `f` before the quote lets you embed expressions inside `{}`:

````python
f"Hello, {name}! Next year you will be {age + 1}."
````

## 4. Expression vs statement

This trips up beginners. Memorize it now.

- **Expression** = code that *produces a value*. `2 + 2`, `len("hello")`, `name.upper()`.
- **Statement** = code that *does something*. `name = "Prince"`, `if x: ...`, `print(x)`.

An expression can be on the right side of `=`. A statement cannot.

Notebooks show the value of the **last expression** in a cell. They don't show anything for statements unless you explicitly print.

````python
x = 5         # statement — assigns 5 to x, no output
````

````python
x             # expression — produces 5
````

````python
print(x)      # statement — does something (prints) but has no output value
````

Notice the difference: the bare `x` cell showed `5` as the *result* (in `Out[]`). The `print(x)` cell showed `5` as *output stream* (no `Out[]`).

It's subtle but it matters when you start writing functions.

## 5. Numbers — int vs float

Python has two kinds of numbers:

- `int` — whole numbers. No size limit. `42`, `-7`, `1_000_000` (underscores for readability).
- `float` — decimals. IEEE 754, so imprecise. `3.14`, `-0.5`, `1.0`.

**Math operators:**

| Operator | Meaning | Example |
|----------|---------|---------|
| `+`      | add | `2 + 3` → `5` |
| `-`      | subtract | `7 - 4` → `3` |
| `*`      | multiply | `2 * 3` → `6` |
| `/`      | divide (always float) | `7 / 2` → `3.5` |
| `//`     | floor divide (integer result) | `7 // 2` → `3` |
| `%`      | modulo (remainder) | `7 % 2` → `1` |
| `**`     | exponent | `2 ** 10` → `1024` |

Run each below:

````python
7 / 2     # always returns a float, even when result is whole
````

````python
7 // 2    # floor division — drops the remainder
````

````python
0.1 + 0.2    # this is not 0.3 — welcome to floating point
````

> ⚠️ **Float weirdness is real.** `0.1 + 0.2 != 0.3`. This isn't a Python bug — it's how computers represent decimals in binary. When you need exact decimals (money!), use the `Decimal` type. For everything else, just be aware.

## 6. Reading error messages

You will get errors. Reading the error is half the battle. Most beginners scan past them; senior engineers read them carefully.

**Run the cell below.** It will error on purpose. Don't panic — read the message.

````python
5 + "hello"
````

Read the error from bottom to top:

1. **The bottom line is the actual error:** `TypeError: unsupported operand type(s) for +: 'int' and 'str'`. Translation: "I don't know how to `+` a number and a text."
2. **The line above shows where:** Cell In[12], line 1.
3. **The arrow `--->`** points to the offending line.

Errors are the language being **honest** with you. They tell you what's wrong, where, and often hint at how to fix it. The fix here is: convert the number to a string first.

**Try it:** Modify the cell to be `str(5) + " hello"`. Re-run. Now it works.

## Self-check

<details>
<summary>1. What does <code>=</code> mean in Python?</summary>

Assignment, not equality. `age = 16` means "the name `age` now refers to the value `16`." Test equality with `==`.
</details>

<details>
<summary>2. What's the difference between an expression and a statement?</summary>

An expression *evaluates* to a value (`2 + 2`, `name`, `f"hi {name}"`). A statement *does* something (`name = "Prince"`, `print(x)`, `if x: ...`). Expressions can be on the right of `=`; statements can't.
</details>

<details>
<summary>3. Why does <code>0.1 + 0.2</code> not equal <code>0.3</code> in Python?</summary>

Floats are stored in binary IEEE 754 format. Some decimals don't have exact binary representations. `0.1 + 0.2` is actually `0.30000000000000004`. For money or exact decimals, use `decimal.Decimal`.
</details>

<details>
<summary>4. What does this error mean? <code>TypeError: unsupported operand type(s) for +: 'int' and 'str'</code></summary>

You tried to `+` a number and a string. Python won't auto-convert. Wrap the number with `str()` or the string with `int()`, whichever makes sense.
</details>

<details>
<summary>5. How is a Jupyter notebook different from a <code>.py</code> file?</summary>

Notebooks are interactive — cells run one at a time, state persists between cells, results display inline. Great for exploration. `.py` files are scripts — run top to bottom, no inline output, deploys to production. You'll use both: notebooks to try things, scripts to ship them.
</details>
