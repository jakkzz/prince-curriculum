## The mental model

Control flow is *the order in which lines run*. Without it, Python reads line 1, then line 2, then line 3. With it, you can:

- **Branch:** "if condition, do X; else, do Y" (`if`/`elif`/`else`)
- **Repeat:** "for each item in this collection, do X" (`for`)
- **Loop until:** "keep doing X while condition is true" (`while`)
- **Skip / stop:** `continue` skips this iteration; `break` exits the loop entirely

> 💡 **In the wild:** Every web request your phone makes uses control flow. Netflix's recommendation engine is *millions* of `if` and `for` and `while` statements composed together. Today you learn the building blocks for everything else.

## 1. `if` / `elif` / `else`

Branch on a condition. Indentation defines what's inside each branch (Python uses indentation, not braces).

```python
if condition:
    # runs if condition is True
elif other_condition:
    # runs if condition is False AND other_condition is True
else:
    # runs if all the above are False
```

Conditions are anything that evaluates to truthy or falsy.

````python
age = 16

if age >= 18:
    print("You're an adult.")
elif age >= 13:
    print("You're a teenager.")
else:
    print("You're a kid.")
````

**Try it:** Change `age` to 25, then 8, then 13. Re-run each time and watch which branch fires.

### Comparison operators

| Operator | Meaning |
|----------|---------|
| `==`     | equal to (NOT a single `=` — that's assignment) |
| `!=`     | not equal to |
| `<` `>`  | less than, greater than |
| `<=` `>=` | less or equal, greater or equal |
| `in`     | membership: `"a" in "banana"` is True |
| `not in` | the opposite of `in` |
| `is`     | identity: `x is None` |

### Logical operators (combining conditions)

| Operator | Meaning |
|----------|---------|
| `and` | both must be True |
| `or` | either is True |
| `not` | flips True/False |

````python
word = "thorough"
is_long = len(word) >= 7
has_th = "th" in word

if is_long and has_th:
    print(f"{word!r} is a long word with 'th'.")
elif is_long or has_th:
    print(f"{word!r} matches at least one criterion.")
else:
    print(f"{word!r} matches neither.")
````

> ⚠️ **Common bug:** Using `=` instead of `==` in a condition. `if x = 5:` is a syntax error; Python catches this. But `if x == y:` vs `if x = y:` can be easy to typo. Read your `if` lines carefully.

## 2. `for` loops — iterate over things

The most common loop in Python. Reads naturally: "for each thing in this collection, do this."

Works on anything *iterable*: lists, tuples, strings, dicts, sets, ranges, files…

````python
words = ["thorough", "ubiquitous", "nuance"]

for word in words:
    print(f"{word}: {len(word)} letters")
````

### `range()` — generate a sequence of numbers

Use when you need to repeat N times or generate a number sequence.

````python
for i in range(5):
    print(i)
````

````python
# range(start, stop, step)
for i in range(10, 0, -2):
    print(i)
````

### `enumerate()` — when you want the index AND the value

Beginners often write `for i in range(len(words)):` then `words[i]`. **Don't.** Use `enumerate()`:

````python
words = ["thorough", "ubiquitous", "nuance"]

for i, word in enumerate(words, start=1):
    print(f"{i}. {word}")
````

### Iterating dicts

Dicts give you keys by default. Use `.items()` for key + value pairs.

````python
word_entry = {"word": "thorough", "ipa": "/ˈθʌrə/", "thai": "ละเอียด"}

for key, value in word_entry.items():
    print(f"{key:6} : {value}")
````

## 3. `while` loops — repeat until done

Use when you don't know how many times you'll loop in advance. Common for: user input loops, retry logic, game loops.

````python
countdown = 5

while countdown > 0:
    print(f"T-minus {countdown}")
    countdown -= 1

print("Liftoff!")
````

> ⚠️ **Infinite loop danger.** If the condition never becomes False, your program never stops. Always make sure *something* in the loop body changes the condition. (For runaway loops in a terminal: Ctrl-C kills them. In a notebook: Stop the kernel.)

## 4. `break` and `continue`

Two ways to short-circuit a loop:

- **`break`** — exit the loop entirely, even if more iterations remain.
- **`continue`** — skip the rest of this iteration, go to the next one.

### `break` example: stop on first match

````python
words = ["thorough", "ubiquitous", "nuance", "resilient"]
target = "nuance"

for word in words:
    if word == target:
        print(f"Found {word!r}!")
        break
    print(f"  not {word}…")
````

### `continue` example: skip odd-length words

````python
words = ["thorough", "ubiquitous", "nuance", "a"]

for word in words:
    if len(word) % 2 == 1:
        continue  # skip — go to next word
    print(f"{word} has even length ({len(word)})")
````

## 6. List comprehensions — the Pythonic loop

A compact way to build a list from a loop. The pattern:

```python
[expression for item in iterable if condition]
```

Reads as: "the expression, for each item in iterable, where condition is true."

````python
# Long way
squares_long = []
for x in range(10):
    squares_long.append(x * x)

# Pythonic way
squares = [x * x for x in range(10)]

squares == squares_long
````

````python
# With a filter
evens_squared = [x * x for x in range(10) if x % 2 == 0]
evens_squared
````

````python
# Useful for the english-helper project
words = ["thorough", "ubiquitous", "nuance", "resilient", "a"]
long_words = [w for w in words if len(w) >= 7]
long_words
````

There are also **dict comprehensions** and **set comprehensions** with the same idea:

````python
# Dict comprehension: word → length
lengths = {w: len(w) for w in ["thorough", "ubiquitous", "nuance"]}
lengths
````

> 💡 **Rule of thumb:** if a comprehension would have to be longer than 1 line or wrap awkwardly, use a regular `for` loop. Comprehensions are for clear, simple transformations, not for showing off.

```python
# pronunciation_quiz.py
import random

WordEntry = dict[str, str]

WORDS: list[WordEntry] = [
    {"word": "thorough",   "ipa": "/ˈθʌrə/",        "thai": "ละเอียด"},
    {"word": "ubiquitous", "ipa": "/juːˈbɪkwɪtəs/", "thai": "พบเห็นได้ทั่วไป"},
    {"word": "nuance",     "ipa": "/ˈnuːɑːns/",     "thai": "ความละเอียดอ่อน"},
    {"word": "resilient",  "ipa": "/rɪˈzɪliənt/",   "thai": "ยืดหยุ่น"},
    {"word": "inevitable", "ipa": "/ɪˈnevɪtəbl/",   "thai": "ที่ไม่อาจหลีกเลี่ยงได้"},
]


def main() -> None:
    correct = 0
    total = 0
    print("\n🎯 English Helper — Pronunciation Quiz\n")
    while True:
        entry = random.choice(WORDS)
        answer = input(f"Type the IPA for {entry['word']!r} (or 'quit'): ").strip()
        if answer.lower() == "quit":
            break
        total += 1
        if answer == entry["ipa"]:
            print("✓ Correct!\n")
            correct += 1
        else:
            print(f"✗ Not quite. Correct IPA: {entry['ipa']}\n")
    if total == 0:
        print("You didn't try any. See you next time.")
    else:
        pct = 100 * correct / total
        print(f"Final score: {correct}/{total} ({pct:.0f}%)")


if __name__ == "__main__":
    main()
```

## Self-check

<details>
<summary>1. What's the difference between <code>=</code> and <code>==</code>?</summary>

`=` assigns. `==` compares. Using `=` where you meant `==` is a classic bug in C; in Python it's a syntax error in conditions, so the language saves you.
</details>

<details>
<summary>2. What does <code>break</code> do? What about <code>continue</code>?</summary>

`break` exits the loop entirely. `continue` skips the rest of this iteration and starts the next one.
</details>

<details>
<summary>3. Why use <code>enumerate(words)</code> instead of <code>for i in range(len(words))</code>?</summary>

Cleaner, less error-prone, more Pythonic. `enumerate()` gives you `(index, item)` pairs directly; the range-and-index pattern leaks implementation detail and is one off-by-one away from a bug.
</details>

<details>
<summary>4. What's a list comprehension? When should you NOT use one?</summary>

Compact `[expr for item in iterable if condition]` syntax for building lists. Don't use them when: the expression is complex, you'd need nested comprehensions, or the result would be longer than one screen-line. Clarity > cleverness.
</details>

<details>
<summary>5. Why doesn't <code>input()</code> work well in a notebook?</summary>

Notebooks expect cells to run to completion. `input()` blocks waiting for typing. It technically works in Jupyter (you'll see a prompt) but it's awkward — better to put interactive code in a `.py` file and run it from the terminal.
</details>
