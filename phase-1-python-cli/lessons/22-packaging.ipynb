## What you'll do today

## 1. The entry point

````python
# pyproject.toml
[project]
name = 'english-helper'
version = '0.1.0'
description = 'A spaced-repetition CLI for English vocabulary'
readme = 'README.md'
requires-python = '>=3.13'
dependencies = [
    'requests>=2.31',
]

[project.scripts]
english-helper = 'english_helper.cli:main'

[build-system]
requires = ['hatchling']
build-backend = 'hatchling.build'

[tool.hatch.build.targets.wheel]
packages = ['src/english_helper']
````

`english-helper = 'english_helper.cli:main'` — when the tool is installed, the shell command `english-helper` will invoke `english_helper.cli.main()`.

**Dashes vs underscores:** the shell command (left of `=`) can have dashes. The import path (right) uses underscores. Don't conflate them.

## 2. Install globally with `uv tool install`

````python
# From inside your english-helper repo:
# uv tool install --from . english-helper
# 
# Now:
# english-helper --help
# english-helper add thorough
# english-helper list
#
# Works from any directory.
````

`uv tool install` creates an isolated environment for the tool, installs your package, and puts the entry-point script in `~/.local/bin` (or similar). No need to be in the project folder.

**To update after code changes:**

```bash
uv tool install --from . english-helper --force
```

(or `uv tool uninstall english-helper && uv tool install --from . english-helper`)

## 3. README + LICENSE

**`README.md`** — what a user sees on GitHub. Required sections:

1. **What it does** (1 sentence)
2. **Install** (`uv tool install --from . english-helper`)
3. **Quickstart** (one example showing the killer feature)
4. **Commands** (cheat sheet table)
5. **Where data lives** (`~/.english-helper/`)
6. **License**

**`LICENSE`** — pick MIT for personal/learning projects. One file, ~20 lines.

## 4. Versioning + releases

Use [semver](https://semver.org/) — `MAJOR.MINOR.PATCH`:

- **PATCH** (0.1.0 → 0.1.1): bug fix, no API change
- **MINOR** (0.1.0 → 0.2.0): new feature, backward compatible
- **MAJOR** (0.1.0 → 1.0.0): breaking change

**v0.1.0 today.** Tag it:

```bash
git tag -a v0.1.0 -m 'First release'
git push origin v0.1.0
```

**v1.0.0 is when the tool is solid and you'd be embarrassed to break the API.** Don't rush it.

## 5. Publishing to PyPI (optional)

Make your tool installable by **anyone** with `pip install english-helper`:

1. Create an account on [pypi.org](https://pypi.org).
2. `uv publish` — uploads the wheel.
3. (Optional) Set up trusted publishing in CI so releases auto-publish.

**Don't publish a name you might want.** PyPI names are first-come-first-served. Once `english-helper` is taken, you can't have it. For learning projects: prefix with your username (`prince-english-helper`).
