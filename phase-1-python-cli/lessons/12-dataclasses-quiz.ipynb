## Section B — Predict the output

**B1.**

```python
from dataclasses import dataclass
@dataclass
class P: x: int; y: int
p = P(1, 2)
print(p)
```

**B2.**

```python
from dataclasses import dataclass
@dataclass
class P: x: int; y: int
print(P(1, 2) == P(1, 2))
```

**B3.**

```python
from dataclasses import dataclass
@dataclass(frozen=True)
class P: x: int
p = P(1)
p.x = 2
```

**B4.**

```python
from dataclasses import dataclass, field
@dataclass
class C: items: list = field(default_factory=list)
c1 = C(); c2 = C(); c1.items.append(1); print(c2.items)
```

**B5.**

```python
from dataclasses import dataclass
@dataclass(order=True)
class S: score: int; name: str
xs = [S(5, 'a'), S(3, 'b')]
print(sorted(xs))
```

**B6.**

```python
from dataclasses import dataclass
@dataclass(frozen=True)
class P: x: int
print(hash(P(1)))
```

**B7.**

```python
from dataclasses import dataclass
@dataclass
class P: x: int
print(hash(P(1)))
```

**B8.**

```python
from dataclasses import dataclass
@dataclass
class C:
    x: int
    def double(self): return self.x * 2
print(C(5).double())
```
