## Section B — Predict the output

**B1.**

```python
class C:
    def __init__(self, n): self.n = n
    @property
    def doubled(self): return self.n * 2
print(C(5).doubled)
```

**B2.**

```python
class C:
    @property
    def x(self): return 1
c = C()
c.x = 5
```

**B3.**

```python
class A:
    def hi(self): return 'A'
class B(A):
    def hi(self): return 'B'
print(A().hi(), B().hi())
```

**B4.**

```python
class A:
    def hi(self): return 'A'
class B(A): pass
print(B().hi())
```

**B5.**

```python
class A:
    def __init__(self, x): self.x = x
class B(A):
    def __init__(self, x, y):
        super().__init__(x)
        self.y = y
b = B(1, 2)
print(b.x, b.y)
```

**B6.**

**B7.**

```python
from abc import ABC, abstractmethod
class A(ABC):
    @abstractmethod
    def f(self): ...
class B(A):
    def f(self): return 1
print(B().f())
```

**B8.**

```python
class C:
    def __init__(self): self.children = []
    def add_child(self, c): self.children.append(c)
c = C(); c.add_child('a'); c.add_child('b')
print(c.children)
```
