## Section B — Predict the output

**B1.**

```python
class Cat:
    def __init__(self, name):
        self.name = name
c = Cat('Bella')
print(c.name)
```

**B2.**

```python
class Cat:
    species = 'feline'
    def __init__(self, name):
        self.name = name
c = Cat('Bella')
print(c.species, Cat.species)
```

**B3.**

```python
class C:
    def __init__(self): self.x = 1
    def double(self): return self.x * 2
print(C().double())
```

**B4.**

```python
class C:
    def __init__(self, x): self.x = x
print(C(5) == C(5))
```

**B5.**

```python
class C:
    def __init__(self, x): self.x = x
    def __eq__(self, other): return self.x == other.x
print(C(5) == C(5))
```

**B6.**

```python
class C:
    def __repr__(self): return 'CUSTOM'
print(C())
print(repr(C()))
```

**B7.**

```python
class Counter:
    count = 0
    def __init__(self):
        Counter.count += 1
for _ in range(3): Counter()
print(Counter.count)
```

**B8.**

```python
class C:
    def m(self): return id(self)
c = C()
print(c.m() == id(c))
```
